# DATA TYPES

int -> integers
float -> floating point number(decimal number)

```python
print(2+4)
print(2-4)
print(2*4)
print(2/4)
print(2//4)
print(2%4)
print(2**4)
```

in the above code 2,4 are the integers (int) and (+,-,*,/) are the actions performed on them and we have got the results some in integer form and some in decimal form.

```python
print(type(2+4))
print(type(2-4))
print(type(2*4))
print(type(2/4))
print(type(2//4))
print(type(2%4))
print(type(2**4))
```

if we look at the type of result in the output 
6
-2
8
0.5
0
2
16 
then only 0.5 is float and rest are int data types including zero (0)

the difference between int and float is even though they both are numbers, float takes up a lot of space, since int can be converted into binary numbers but float cannot, hence in a decimal number, number before the decimal and after the decimals are stored seperately, hence they take up a lot of space

```python
# Math Functions
# -> abs() -> Absolute Value
print(abs(-5))
print(abs(5))
```

```python
# -> round() -> Rounding a number to the nearest integer
print(round(3.2))
print(round(3.8))
```

```python
# -> sqrt() -> Square Root
import math
print(math.sqrt(16))
print(math.sqrt(25))
```

```python
# operators precedence

print((5 + 4) * 10 / 2)

print(((5 + 4) * 10) / 2)

print((5 + 4) * (10 / 2))

print(5 + (4 * 10) / 2)

print(5 + 4 * 10 // 2)
```

```python
# Strings
# Concatenation
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print(full_name)
```

```python
# Repetition
magic_word = "love u "
repeat = magic_word * 3000
print(repeat)
```

```python
# counting
count = len(repeat)
print(count)
```

```python
# indexing 

print(full_name[0])  # J
print(full_name[1])  # o
print(full_name[2])  # h
print(full_name[3])  # n
print(full_name[4])  # space
print(full_name[5])  # D
```

```python
# slicing
print(full_name[0:4])  
print(full_name[5:8]) 
print(full_name[:4])  
print(full_name[:])    
print(full_name[::-1]) # reverse the string
```

```python
# string methods
print(full_name.upper())
print(full_name.lower())
print(full_name.title())
print(full_name.replace("John", "Jane"))
print(full_name.split())  
```

```python
name = 'mothilal'
print(name.split('l'))
print(name.count('l'))
```

```python
# formatted strings
age = 23
print(f"My name is {name} and I am {age} years old.")

print("My name is {} and I am {} years old.".format(name, age))
```
