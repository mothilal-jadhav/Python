```python
#oops

class PlayerCharacter:
    def __init__(self,name,age): #constructor method / init method, auto assigns the name and age to the instance of the class
        self.name = name #attribute of the class, self is the instance of the class
        self.age = age #attribute of the class, self is the instance of the class

    def run(self): #method of the class, self is the instance of the class, run is the method that will be called when the player runs
        return "run" #returning the string "run" when the method is called

player1 = PlayerCharacter("Mothilal", 25) #creating an instance of the class, player1 is the instance of the class, "Mothilal" is the name and 25 is the age
print(player1.name)
print(player1.age)
print(player1.run())
```
